# Etapa 1 — Criação do Bucket e Upload do Arquivo para o S3

Este notebook realiza a criação de um bucket S3 e envia o arquivo original (`nucleo_seguranca.csv`) para o serviço da AWS.

As credenciais devem ser inseridas manualmente para fins de autenticação temporária, conforme instruído no desafio.

```python
# 📦 Bibliotecas necessárias
import boto3
from botocore.exceptions import ClientError, NoCredentialsError
```

```python
# 🌍 Região e nome do bucket
regiao = 'us-east-1'
meu_bucket = 'pb-desafio-leonardo'
```

```python
# 📄 Caminho do arquivo local e nome desejado no S3
caminho_arquivo = 'nucleo_seguranca.csv'
nome_no_s3 = 'dados/nucleo_seguranca.csv'
```

```python
# 🔐 Insira suas credenciais temporárias abaixo
aws_key = 'SUA_AWS_KEY'
aws_secret = 'SEU_AWS_SECRET'
aws_token = 'SEU_AWS_TOKEN_TEMPORARIO'
```

## Função para criação do bucket

```python
def criar_meu_bucket(chave, segredo, token, regiao, nome_bucket):
    try:
        s3 = boto3.client('s3',
                          aws_access_key_id=chave,
                          aws_secret_access_key=segredo,
                          aws_session_token=token,
                          region_name=regiao)

        if regiao == 'us-east-1':
            resposta = s3.create_bucket(Bucket=nome_bucket)
        else:
            resposta = s3.create_bucket(
                Bucket=nome_bucket,
                CreateBucketConfiguration={'LocationConstraint': regiao}
            )

        print(f"✅ Bucket '{nome_bucket}' criado com sucesso.")
        return resposta

    except ClientError as erro:
        print(f"❌ Erro ao criar o bucket: {erro}")
        return None
```

## Função para fazer upload do arquivo ao bucket

```python
def subir_arquivo(chave, segredo, token, bucket, arquivo_local, nome_objeto):
    try:
        s3 = boto3.client('s3',
                          aws_access_key_id=chave,
                          aws_secret_access_key=segredo,
                          aws_session_token=token)

        s3.upload_file(arquivo_local, bucket, nome_objeto)
        print(f"✅ Upload de '{arquivo_local}' para '{bucket}/{nome_objeto}' realizado com sucesso.")

    except FileNotFoundError:
        print(f"❌ Arquivo '{arquivo_local}' não encontrado.")
    except NoCredentialsError:
        print("❌ Credenciais da AWS inválidas ou não configuradas.")
    except Exception as e:
        print(f"⚠️ Erro inesperado: {e}")
```

## Execução final — Criar bucket e enviar arquivo

```python
resposta_bucket = criar_meu_bucket(aws_key, aws_secret, aws_token, regiao, meu_bucket)
if resposta_bucket:
    subir_arquivo(aws_key, aws_secret, aws_token, meu_bucket, caminho_arquivo, nome_no_s3)
```
